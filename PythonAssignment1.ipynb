## 1.String Concatenation

```python
str1 = "Hello "
name =input("Enter your Name: ")
result =str1+name
print(result)
```

```python
str3 = ", welcome to Python programming"
final_result = result+str3
print(final_result)
```

# 2. String Slicing and Indexing

```python
#first charecter of the string
print("First charecter: ", final_result[0])

#Print the last character of the string.
print("Last charecter: ", final_result[-1])

#Print the first 5 characters of the string.
print("First 5 charecters: ", final_result[:5])

#Print the last 11 characters of the string. 
print("Last 11 charecters: ", final_result[-11:])

#Print the string in reverse. 
print("Reverse string: ", final_result[::-1])

#Use slicing and print the word “Python” from the existing string.
print("Python:", final_result[24:30])
```

## 3. String Methods: 

```python
strM = "Python beginner tutorial"

#Convert the sentence to uppercase. 
print("Uppercase: ", strM.upper())

#Convert the sentence to lowercase.
print("Lowercase: ", strM.lower())

#Use Capitalize and return the sentence to the original input form.
print("Capitalize:", strM.capitalize())

#Count the total number of occurrences of character ‘t’  in the string. 
print("Count of t: ", strM.count('t'))

#Replace all occurrences of “Python” with “Machine Learning”  in the input string strM = “Python beginner tutorial”
print("Replace: ", strM.replace("Python","Machine Learning"))
```

## Tuples (Creation, Modification and Access) :  

```python
#Create 1st tuple with values -> (10, 20, 30), 2nd tuple with values -> (40, 50, 60):  
#first tuple
t1 = (10,20,30)
#second tuple
t2=(40,50,60)

# Concatenate the two tuples and store it in “t_combine” 
t_combine=t1+t2
print("Combine tuple: ", t_combine)

# Repeat the elements of “t_combine” 3 times  
print("Repeated tuple: ", t_combine * 3)

#Access the 3rd element from “t_combine”  
print("Third element: ",t_combine[2])

# Access the first three elements from “t_combine”  
print("First three elements: ", t_combine[:3])

# Access the last three elements from “t_combine”
print("Last three elements: ", t_combine[-3:])
```
